# Berkeley

In 1973, the admission data in UC Berkeley graduate schools showed a clear gender bias: a larger percentage of male applicants was being admitted. However, when the admission rates in each major are analysed, this gender bias is not observed. This is a typical example of [Simpson's paradox](https://en.wikipedia.org/wiki/Simpson%27s_paradox). To understand it, we must calculate the admission rates in total and by gender for each major and in total, as well as the percentage in which the students of each gender apply to the different majors.

```python
%apl d←(↑1⌷∘⍉¨1∘↓)⊃(dd h)←⎕CSV⍠1(⊂⍪⍬ ⍬ 1⍨)'csv/berkeley.csv'
df = pd.read_csv("csv/berkeley.csv")
display(df)
%apl ⍕(''⊂⍛⍪h)⍪5(↑⍪'...'⊂⍛⍪-⍛↑)⍳∘≢⍛(⍪⍤¯1)⊃⍪⍤¯1⌿⊂⍤¯1¨dd
```

```python
%apl a←⊂∘⍋⍛⌷((⍉1 2⊂⍛⌷⊢)⍪∘(≢⍪⍨'A'+⌿⍤=⊢)⌸3∘⌷)d ⋄
adm = ('Admission', lambda c:(c=='Accepted').sum())
app = ('Admission', 'count')
a = df.groupby(['Major','Gender']).agg(Admitted=adm, Applicants=app)
display(a)
%apl ⍕(ha←(2 3⊂⍛⌷h)⍪'Admitted' 'Applicants')⍪a
```

```python
%apl g←a⍪'T'⍪⍤¯1(2∘⌷⍪∘(+⌿)⌸∘⍉3 4⊂⍛⌷⊢)⍉a ⋄
gt = pd.concat([a.groupby('Gender').sum()], keys=['Total'], names=['Major'])
g = pd.concat([a, gt])
display(g)
%apl ⍕ha⍪g
```

```python
%apl m←⊂∘⍋⍛⌷(⍉⍪1∘⌷⍪∘('T'⍪+⌿)⌸∘⍉3 4⊂⍛⌷⊢)⍉g ⋄
mt = pd.concat([g.groupby('Major').sum()], keys=['T'], names=['Gender']).reorder_levels([1, 0])
m = pd.concat([g, mt]).sort_index()
display(m)
%apl ⍕ha⍪m
```

```python
%apl ar←(⍉⍪(⍤¯1)100×3∘⌷÷4∘⌷)⍉m ⋄
ar = m.assign(PctAdmitted=100*m.Admitted/m.Applicants)
display(ar)
%apl ⍕(har←ha⍪⊂'%Admitted')⍪ar
```

```python
ar = m.copy()
ar['PctAdmitted'] = 100*ar.Admitted/ar.Applicants
ar
```

```python
%apl mr←ar⍪⍤¯1(100×⊢÷≢⍴¯3∘↑)4⌷⍉ar ⋄
mr = ar.assign(PctApplicants = 100*ar.Applicants/ar.loc['Total']['Applicants'])
display(mr)
%apl ⍕(hmr←har⍪⊂'%Applicants')⍪mr
```

```python
# read data
%apl d←(↑1⌷∘⍉¨1∘↓)⊃(dd h)←⎕CSV⍠1(⊂⍪⍬ ⍬ 1⍨)'csv/berkeley.csv'
df = pd.read_csv("csv/berkeley.csv")

# group by gender and by major
%apl a←⊂∘⍋⍛⌷((⍉1 2⊂⍛⌷⊢)⍪∘(≢⍪⍨'A'+⌿⍤=⊢)⌸3∘⌷)d ⋄
adm = ('Admission', lambda c:(c=='Accepted').sum())
app = ('Admission', 'count')
a = df.groupby(['Major','Gender']).agg(Admitted=adm, Applicants=app)

# totals by gender and by major
%apl g←a⍪'T'⍪⍤¯1(2∘⌷⍪∘(+⌿)⌸∘⍉3 4⊂⍛⌷⊢)⍉a ⋄
%apl m←⊂∘⍋⍛⌷(⍉⍪1∘⌷⍪∘('T'⍪+⌿)⌸∘⍉3 4⊂⍛⌷⊢)⍉g ⋄
gt = pd.concat([a.groupby('Gender').sum()], keys=['Total'], names=['Major'])
g = pd.concat([a, gt])
mt = pd.concat([g.groupby('Major').sum()], keys=['T'], names=['Gender']).reorder_levels([1, 0])
m = pd.concat([g, mt]).sort_index()

# admission and applicants ratios
%apl ar←(⍉⍪(⍤¯1)100×3∘⌷÷4∘⌷)⍉m ⋄
%apl mr←ar⍪⍤¯1(100×⊢÷≢⍴¯3∘↑)4⌷⍉ar ⋄
ar = m.assign(PctAdmitted=100*m.Admitted/m.Applicants)
mr = ar.assign(PctApplicants = 100*ar.Applicants/ar.loc['Total']['Applicants'])
```

# Iris

The [Iris dataset](https://en.wikipedia.org/wiki/Iris_flower_data_set) contains measures of different features for three different classes of Iris flowers. This dataset is frequently used as an example for different data classification techniques.

```python
%apl d←(⊃⍪⍥⊂∘⍉∘↑1∘↓)¯1⊖⎕CSV⍠1(⊂⍪⍬(1⍪⍨4⍴2)⍨)'csv/iris.csv'
%apl h←'sl' 'sw' 'pl' 'pw'
df = pd.read_csv("csv/iris.csv", header=None, names=['sl', 'sw', 'pl', 'pw', 'class'])
display(df)
%apl ⍕('' 'class')h⍪¨5(↑⍪'...'⊂⍛⍪-⍛↑)¨(⍳∘≢⍪⍤¯1⍪)∘↓¨@1⊢d
```

```python
%apl _a←{((↓∘⊃⍪⍤1∘⍺⍺⌸2∘⊃)⍪'Total'⊂⍛⍪⍤1∘⍺⍺ 2∘⊃)⍵} ⋄
def A(df, a):
    t = pd.concat([df[['sl', 'sw', 'pl', 'pw']].agg(a)], keys=['Total'], names=['class']).unstack()
    return pd.concat([df.groupby('class').agg(a), t])
```

```python
%apl std←(2*∘÷⍨+⌿÷¯1+≢)2*⍨⊢-⍤1+⌿÷≢ ⋄
%apl s←(⌊⌿⍪(+⌿÷≢)⍪std⍪⌈⌿)_a d
s = [A(df, x) for x in ('min', 'mean', 'std', 'max')]
[display(x) for x in s]
%apl {⍕⍵⊣⎕PP←3}('class'⊂⍛⍪,'⌊AS⌈'∘.⍪','(≠⊆⊢)'sl,sw,pl,pw')⍪s
```

```python
%apl pct←{((2÷⍨+⌿)⊢⌷⍨∘⊂⍋⌷⍨∘⊂∘⌈100÷⍨⍺×0 1+≢)⍵} ⋄
%apl q←(,25 50 75∘.pct↓∘⍉)_a d
q = [A(df, lambda d: d.quantile(q=x)) for x in [0.25, 0.50, 0.75]]
[display(x) for x in q]
%apl ⍕('class'⊂⍛⍪,⍉25 50 75⍕¨⍛(∘.⍪⍨)h)⍪q
```

```python
A(df, 'corr')
```

```python
%apl pcc←+.×⍥((⊢÷2*∘÷⍨+.×⍨)⊢-+⌿÷≢) ⋄
%apl cm←⊂⍤¯1(∘.pcc⍨↓∘⍉)_a d
mt = pd.concat([df.corr(numeric_only=True)], keys=['Total'], names=['class'])
cm = pd.concat([df.groupby('class').corr(), mt])
display(cm)
%apl ⍕('class'⊂⍛⍪h)⍪⊃⍪⌿cm
```

```python
%apl c←cm⍪⊂(⊂'class')⍪({⊂⍳⍨↓⊃⍵}pcc¨{↓⍉2⊃⍵})d
ct = df.corrwith(pd.Series(pd.factorize(df['class'])[0]), numeric_only=True)
cc = pd.concat([ct], keys=[('Total', 'class')], names=['class', '']).unstack()
c = pd.concat([cm, cc])
display(c)
%apl ⍕('class'⊂⍛⍪h)⍪⊃⍪⌿c
```

```python
# read data
%apl d←(⊃⍪⍥⊂∘⍉∘↑1∘↓)¯1⊖⎕CSV⍠1(⊂⍪⍬(1⍪⍨4⍴2)⍨)'csv/iris.csv'
%apl h←'sl' 'sw' 'pl' 'pw'
df = pd.read_csv("csv/iris.csv", header=None, names=['sl', 'sw', 'pl', 'pw', 'class'])

# aggregate with total
%apl _a←{((↓∘⊃⍪⍤1∘⍺⍺⌸2∘⊃)⍪'Total'⊂⍛⍪⍤1∘⍺⍺ 2∘⊃)⍵} ⋄
def A(df, a):
    t = pd.concat([df[['sl', 'sw', 'pl', 'pw']].agg(a)], keys=['Total'], names=['class']).unstack()
    return pd.concat([df.groupby('class').agg(a), t])

# summary, percentiles and correlation
%apl std←(2*∘÷⍨+⌿÷¯1+≢)2*⍨⊢-⍤1+⌿÷≢ ⋄
%apl s←(⌊⌿⍪(+⌿÷≢)⍪std⍪⌈⌿)_a d
s = [A(df, x) for x in ('min', 'mean', 'std', 'max')]
%apl pct←{((2÷⍨+⌿)⊢⌷⍨∘⊂⍋⌷⍨∘⊂∘⌈100÷⍨⍺×0 1+≢)⍵} ⋄
%apl q←(,25 50 75∘.pct↓∘⍉)_a d
q = [A(df, lambda d: d.quantile(q=x)) for x in [0.25, 0.50, 0.75]]
%apl pcc←+.×⍥((⊢÷2*∘÷⍨+.×⍨)⊢-+⌿÷≢) ⋄
%apl cm←⊂⍤¯1(∘.pcc⍨↓∘⍉)_a d
mt = pd.concat([df.corr(numeric_only=True)], keys=['Total'], names=['class'])
cm = pd.concat([df.groupby('class').corr(), mt])
%apl c←cm⍪⊂(⊂'class')⍪({⊂⍳⍨↓⊃⍵}pcc¨{↓⍉2⊃⍵})d
ct = df.corrwith(pd.Series(pd.factorize(df['class'])[0]), numeric_only=True)
cc = pd.concat([ct], keys=[('Total', 'class')], names=['class', '']).unstack()
c = pd.concat([cm, cc])
```

# Google

A [time series](https://en.wikipedia.org/wiki/Time_series) is a series of data points indexed by time. As an example, we can analyse the number of searches for "scotch" in the last 5 years taken from google trends: https://trends.google.com/trends/explore?date=today%205-y&q=scotch

```python
%apl (d h)←↑¨@1 ⎕CSV⍠1(⊂⍪'N'(4⍴2)1⍨)'-'⎕R','⊢',,'∘⍪¨@1⊢2↓⊃⎕NGET(⊂⍪1⍨)'csv/google-scotch.csv'
df = pd.read_csv("csv/google-scotch.csv", header=1)
display(df)
%apl ⍕(''⊂⍛⍪h)⍪5(↑⍪'...'⊂⍛⍪-⍛↑)⍳∘≢⍛(⍪⍤¯1)⍉d
```

```python
d = df.assign(Week = pd.to_datetime(df['Week'])).set_index('Week')
d
```

```python
%apl t←((⍉2∘↑)⍪∘(+⌿)⌸4∘⌷)d
a = d.groupby([d.index.year, d.index.month]).agg(Total=(d.columns[0], 'sum'))
a.index.names = ['Year', 'Month']
display(a)
%apl ⍕'year' 'month' 'total'⍪5(↑⍪'...'⊂⍛⍪-⍛↑)t
```

```python
%apl l←'min' 'avg' 'max' 'total' 'change'
def A(df, by):
    (m, a, M, t) = [('Total',x) for x in ['min', 'mean', 'max', 'sum']]
    return df.groupby(by).agg(min=m, mean=a, max=M, total=t)
```

```python
%apl s←(2∘↑((⊂⊂∘⍋⍛⌷)⍪∘(⌊⌿⍪(+⌿÷≢)⍪⌈⌿⍪+⌿)⌸)(⍤1)3∘⌷)⍉t
%apl my←⊖0 ¯1∘↓¨@2⊢'year' 'month'((⊣⊂⍛⍪l⍨)⍪⊢⍪(⍤¯1)'⍬'⍪2-⍨⌿5⌷⍉⍤⊢)¨s
m = A(a, 'Month')
display(m)
%apl ⍕my
```

```python
y = A(a, 'Year')
y = y.assign(change=y['total'].diff())
y
```

```python
%apl F←{(⍉⊢(⍴⍤⊢⍴⍺⊣(@1)3⊃¨@(~zi)∘((+⌿¯1∘↓)¨@(zi←∨⌿⍤1↑∊0⍨)),⍤⊢⍪¨⊣((⍺⍺ 3⌷⍉)⍪'⍬'⍨)⍛(⌷⍨)∘⊂0 ¯1∘↓⍛⍳∘↑∘,)¯1+(12+1)⍳⍛(∘.⍪⍨)1+0⍪∘∪1⌷⍉)⍵} ⋄
%apl tc←(((⊂'Total')⊢F⊢)⍪(⍤¯1)(⊂'Chnage')('⍬'⍪2-⍨⌿⊢)F⊢)t
tm = a.unstack(level=0)
display(tm)
%apl ⍕tc
```

```python
cm = a.diff().unstack(level=0)
cm
```

```python
%apl (d h)←↑¨@1 ⎕CSV⍠1(⊂⍪'N'(4⍴2)1⍨)'-'⎕R','⊢',,'∘⍪¨@1⊢2↓⊃⎕NGET(⊂⍪1⍨)'csv/google-scotch.csv'
df = pd.read_csv("csv/google-scotch.csv", header=1)
d = df.assign(Week = pd.to_datetime(df['Week'])).set_index('Week')
%apl t←((⍉2∘↑)⍪∘(+⌿)⌸4∘⌷)d
a = d.groupby([d.index.year, d.index.month]).agg(Total=(d.columns[0], 'sum'))
a.index.names = ['Year', 'Month']
%apl l←'min' 'avg' 'max' 'total' 'change'
def A(df, by):
    (m, a, M, t) = [('Total',x) for x in ['min', 'mean', 'max', 'sum']]
    return df.groupby(by).agg(min=m, mean=a, max=M, total=t)
%apl s←(2∘↑((⊂⊂∘⍋⍛⌷)⍪∘(⌊⌿⍪(+⌿÷≢)⍪⌈⌿⍪+⌿)⌸)(⍤1)3∘⌷)⍉t
%apl my←⊖0 ¯1∘↓¨@2⊢'year' 'month'((⊣⊂⍛⍪l⍨)⍪⊢⍪(⍤¯1)'⍬'⍪2-⍨⌿5⌷⍉⍤⊢)¨s
m = A(a, 'Month')
y = A(a, 'Year')
y = y.assign(change=y['total'].diff())
%apl F←{(⍉⊢(⍴⍤⊢⍴⍺⊣(@1)3⊃¨@(~zi)∘((+⌿¯1∘↓)¨@(zi←∨⌿⍤1↑∊0⍨)),⍤⊢⍪¨⊣((⍺⍺ 3⌷⍉)⍪'⍬'⍨)⍛(⌷⍨)∘⊂0 ¯1∘↓⍛⍳∘↑∘,)¯1+(12+1)⍳⍛(∘.⍪⍨)1+0⍪∘∪1⌷⍉)⍵} ⋄
%apl tc←(((⊂'Total')⊢F⊢)⍪(⍤¯1)(⊂'Chnage')('⍬'⍪2-⍨⌿⊢)F⊢)t
tm = a.unstack(level=0)
cm = a.diff().unstack(level=0)
```
